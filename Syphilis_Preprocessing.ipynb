```python
from google.colab import files
import pandas as pd

# Upload CSV
uploaded = files.upload()

# Read
df = pd.read_csv(next(iter(uploaded)))
df.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Basic Dataset Statistics
print("Dataset Shape:", df.shape)
print("\nTotal Patients:", len(df))
print("\nMissing Values:\n")
print(df.isnull().sum())




#  Violin Plot for Age Distribution

plt.figure(figsize=(12,6))

sns.violinplot(
    x=df["AGE"],
    inner="quartile",
    linewidth=2
)

plt.title("Age Distribution of Patients", fontsize=20)
plt.xlabel("Age", fontsize=15)
plt.show()




# KDE Plot

plt.figure(figsize=(12,6))

sns.kdeplot(
    data=df,
    x="AGE",
    fill=True,
    linewidth=2
)

plt.title("Density Distribution of Patient Age", fontsize=20)
plt.xlabel("Age")
plt.ylabel("Density")
plt.show()


# Fancy Cluster Map

sample_df = df.select_dtypes(include=np.number).sample(100)

sns.clustermap(
    sample_df,
    figsize=(14,12),
    cmap='viridis',
    standard_scale=1
)

plt.suptitle("Hierarchical Feature Clustering", y=1.02)
plt.show()
```

```python
#  Correlation Heatmap

plt.figure(figsize=(18,12))

corr = df.corr(numeric_only=True)

mask = np.triu(np.ones_like(corr, dtype=bool))

sns.heatmap(
    corr,
    mask=mask,
    cmap='coolwarm',
    annot=False,
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": .8}
)

plt.title("Feature Correlation Heatmap", fontsize=22)
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.show()
```

```python

selected_features = [
    'AGE',
    'NUM_RES_HOUSEHOLD',
    'FAM_INCOME',
    'LEVEL_SCHOOLING'
]

sns.pairplot(
    df[selected_features],
    diag_kind='kde',
    corner=True
)

plt.suptitle("Pairwise Relationship Between Clinical Features", y=1.02, fontsize=20)
plt.show()
```
