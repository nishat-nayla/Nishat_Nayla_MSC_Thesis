```python
# ================================
# IMPORT LIBRARIES
# ================================

import os
import zipfile
import cv2
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from collections import Counter

warnings.filterwarnings('ignore')
```

```python
zip_path = "/content/drive/MyDrive/Nishat_Nayla_MSC_Thesis/Datasets/Alzheimer/archive.zip"
```

```python
extract_path = "/content/alzheimer_dataset"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset Extracted Successfully!")
```

```python
dataset_path = "/content/alzheimer_dataset/AugmentedAlzheimerDataset"

classes = os.listdir(dataset_path)

print("Classes Found:")
print(classes)
```

```python
# creating dataframe of image paths and labels

image_paths = []
labels = []

for cls in classes:

    class_path = os.path.join(dataset_path, cls)

    for img_name in os.listdir(class_path):

        img_path = os.path.join(class_path, img_name)

        image_paths.append(img_path)

        labels.append(cls)

df = pd.DataFrame({
    'Image_Path': image_paths,
    'Label': labels
})

print(df.shape)

df.head()
```

```python
# encoding labels and splitting dataset

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

label_encoder = LabelEncoder()

df['Encoded_Label'] = label_encoder.fit_transform(df['Label'])

train_df, test_df = train_test_split(
    df,
    test_size=0.15,
    stratify=df['Encoded_Label'],
    random_state=42
)

train_df, val_df = train_test_split(
    train_df,
    test_size=0.15,
    stratify=train_df['Encoded_Label'],
    random_state=42
)

print("Training Samples   :", len(train_df))
print("Validation Samples :", len(val_df))
print("Testing Samples    :", len(test_df))
```

```python
# image preprocessing and normalization

IMG_SIZE = 224

def preprocess_images(dataframe):

    images = []
    labels = []

    for img_path, label in zip(
        dataframe['Image_Path'],
        dataframe['Encoded_Label']
    ):

        img = cv2.imread(img_path)

        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))

        img = img / 255.0

        images.append(img)

        labels.append(label)

    return np.array(images), np.array(labels)

X_train, y_train = preprocess_images(train_df)

X_val, y_val = preprocess_images(val_df)

X_test, y_test = preprocess_images(test_df)

print("Train Shape :", X_train.shape)
print("Validation Shape :", X_val.shape)
print("Test Shape :", X_test.shape)
```
