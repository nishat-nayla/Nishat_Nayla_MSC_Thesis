```python

# IMPORT LIBRARIES
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

sns.set_style("whitegrid")
```

```python
# DATASET PATH

dataset_path = "desktop/nna272/msc_thesis/breast_cancer/EMBED_dataset"

# LOAD METADATA CSV
metadata_path = os.path.join(dataset_path, "metadata.csv")

df = pd.read_csv(metadata_path)


# BASIC DATASET INFORMATION
print("Dataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())


# DUPLICATE CHECK
print("\nDuplicate Rows:")
print(df.duplicated().sum())


# CLASS DISTRIBUTION

if 'pathology' in df.columns:
    plt.figure(figsize=(8,5))
    sns.countplot(data=df, x='pathology')
    plt.title("Pathology Distribution")
    plt.xticks(rotation=45)
    plt.show()


# BI-RADS DISTRIBUTION

if 'birads' in df.columns:
    plt.figure(figsize=(8,5))
    sns.countplot(data=df, x='birads')
    plt.title("BI-RADS Distribution")
    plt.show()


# BREAST DENSITY DISTRIBUTION

if 'density' in df.columns:
    plt.figure(figsize=(8,5))
    sns.countplot(data=df, x='density')
    plt.title("Breast Density Distribution")
    plt.show()


# AGE DISTRIBUTION

if 'age' in df.columns:
    plt.figure(figsize=(8,5))
    sns.histplot(df['age'], bins=30, kde=True)
    plt.title("Age Distribution")
    plt.xlabel("Age")
    plt.show()
```

```python

# VIEW POSITION ANALYSIS

if 'view' in df.columns:
    plt.figure(figsize=(8,5))
    sns.countplot(data=df, x='view')
    plt.title("Mammography View Distribution")
    plt.show()


# IMAGE SIZE ANALYSIS

image_extensions = ['.png', '.jpg', '.jpeg']

image_paths = []

for root, dirs, files in os.walk(dataset_path):
    for file in files:
        if any(file.lower().endswith(ext) for ext in image_extensions):
            image_paths.append(os.path.join(root, file))

print(f"\nTotal Images Found: {len(image_paths)}")

widths = []
heights = []

for img_path in image_paths[:500]:  # limit for faster execution
    try:
        img = Image.open(img_path)
        widths.append(img.size[0])
        heights.append(img.size[1])
    except:
        pass

print("\nAverage Width:", np.mean(widths))
print("Average Height:", np.mean(heights))


# IMAGE DIMENSION VISUALIZATION


plt.figure(figsize=(8,5))
plt.scatter(widths, heights, alpha=0.5)
plt.xlabel("Width")
plt.ylabel("Height")
plt.title("Image Dimension Distribution")
plt.show()

# SAMPLE IMAGE VISUALIZATION

sample_images = image_paths[:9]

plt.figure(figsize=(12,12))

for i, img_path in enumerate(sample_images):
    try:
        img = Image.open(img_path)

        plt.subplot(3,3,i+1)
        plt.imshow(img, cmap='gray')
        plt.title(os.path.basename(img_path))
        plt.axis('off')

    except:
        continue

plt.tight_layout()
plt.show()


# CORRELATION HEATMAP
numeric_df = df.select_dtypes(include=['int64', 'float64'])

if len(numeric_df.columns) > 1:
    plt.figure(figsize=(10,8))
    sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
    plt.title("Correlation Heatmap")
    plt.show()


# SUMMARY STATISTICS

print("\nSummary Statistics:")
print(df.describe())


# UNIQUE VALUE ANALYSIS


for col in df.columns:
    print(f"\nColumn: {col}")
    print(df[col].nunique())
```
